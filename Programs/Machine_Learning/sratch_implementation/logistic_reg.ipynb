```python
import numpy as np
```

```python
class LogisticRegression:
    def __init__(self, lr = 0.1):
        self.lr = lr

    def _sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X_train , y_train, n_iter = 1000):
        m, n = X_train.shape
        self.bais = 0
        self.weights = np.zeros(n)

        for _ in range(n_iter):
            z = self.bais + (X_train @ self.weights)
            y_pred = self._sigmoid(z)

            db = (1/m) * (y_pred - y_train)
            dw = (1/m) * (X_train.T @ (y_pred - y_train))

            self.bais -= self.lr * db
            self.weights -= self.lr * dw


    def get_probabilities(self, x_test):
        z = self.bais + x_test @ self.weights
        return self._sigmoid(z)

    def predict(self, X_test, thershold=0.5):
        y_predition_val = self.get_probabilities(X_test)
        y_pred_bool = y_predition_val >= thershold
        return y_pred_bool.astype(int)

```

```python
x = np.array([[1, 2], [2, 3], [3, 4], [4, 5]])
y = np.array([0, 0, 1, 1])

model = LogisticRegression()

model.fit(x, y)
y_pred = model.predict(x)
y_pred
```
