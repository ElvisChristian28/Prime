```python
import numpy as np
```

```python
class LinearRegression:
    def __init__(self, lr = 0.001):
        self.lr = lr

    def fit(self, X_train, y_train, n_iter = 1000):
        m, n = X_train.shape
        self.bais = 0
        self.weights = np.zeros(n)

        for _ in range(n_iter):
            self.y_pred = self.bais + np.dot(X_train, self.weights)

            db = (1/m) * np.sum(self.y_pred - y_train)
            dw = (1/m) * np.dot(X_train.T, self.y_pred - y_train)

            self.bais -= self.lr * db
            self.weights -= self.lr * dw

    def predict(self, x_test):
        self.y_predicted_val =  self.bais + np.dot(x_test, self.weights)
        return self.y_predicted_val
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()

model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)
```

```python
class LinearRegressionOLS:
    def __init__(self):
        pass

    def fit(self, X_train, y_train):
        m, n = X_train.shape
        X_train_b = np.c_[np.ones((m, 1)), X_train]
        theta = np.linalg.inv(X_train_b.T @ X_train_b) @ (X_train_b.T @ y_train)

        self.bais = theta[0]
        self.weights = theta[1:]
        

    def predict(self, x_test):
        self.y_predicted_val = self.bais + (x_test @ self.weights)
        return self.y_predicted_val
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegressionOLS()

model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)
```
