```python
import gymnasium as gym
import numpy as np
import random
```

```python
env = gym.make("CliffWalking-v1")
```

```python
gamma = 0.99 
alpha = 0.5
episodes = 1000
epsilon = 0.2
```

```python
Q = np.zeros((48, 4))
```

```python
def epsilon_greedy(state):
    if random.random() < epsilon:
        return env.action_space.sample()
    else:
        return np.argmax(Q[state])
```

```python
for episode in range(episodes):
    # render = (episode  % 150 == 0)
    # if render:
    #     env = gym.make("CliffWalking-v1", render_mode="human")
    #     print(f"Rendering Enviroment for episode: {episode +  1}/ 1000")
    # else:
    #     env = gym.make("CliffWalking-v1")

    env = gym.make("CliffWalking-v1")
    done = False
    state, _ = env.reset()
    action = epsilon_greedy(state)

    total_reward = 0
    episode_len = 0

    while not done:
        next_state, reward, terminated, trucated, _ = env.step(action)
        next_action = epsilon_greedy(next_state)
        done = terminated or trucated

        Q[state, action] += alpha * (reward + gamma * ( Q[next_state, next_action] - Q[state, action] ))

        state = next_state
        action = next_action

        total_reward += reward
        episode_len += 1
    print(f"Episode Number: {episode + 1} | Total Reward: {total_reward} | Episode Length: {episode_len}")
    env.close()
```

```python
env = gym.make("CliffWalking-v1", render_mode="human")
state, _ = env.reset()
done = False
total_reward = 0
episode_len = 0
while not done:
    action = np.argmax(Q[state])
    next_state, reward, terminated, trucated, _ = env.step(action)
    done = trucated or terminated
    state = next_state
    total_reward += reward
    episode_len += 1

print(f" Total Reward: {total_reward} | Episode Length: {episode_len}")
env.close()
```
