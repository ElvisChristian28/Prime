```python
df = pd.read_csv("Data-Files/raw_data.csv")
df
```

```python
df.isnull() # alias df.isna()
df.isna().sum()
```

```python
df.dropna() # axis=1 for colum
```

```python
cleaned_data = df.copy()
cleaned_data["age"] = cleaned_data["age"].fillna(cleaned_data["age"].mean())
cleaned_data
```

```python
cleaned_data["country"] = cleaned_data["country"].ffill()
cleaned_data
```

```python
cleaned_data["country"] = cleaned_data["country"].bfill()
cleaned_data
```

```python
df.duplicated()
```

```python
cleaned_data = cleaned_data.drop_duplicates()
cleaned_data
```

```python
cleaned_data.dtypes
cleaned_data["age"] = cleaned_data["age"].astype("int64").copy()
cleaned_data.dtypes
```

```python
cleaned_data
```

```python
date_str = pd.to_datetime("2025-12-07")
print(type(date_str))
```

```python
date_str = pd.Series([pd.to_datetime("2025-12-07")])
type(date_str.dtypes)
```

```python
aqi_data["timestamp"] = pd.to_datetime(aqi_data["timestamp"]).copy()
aqi_data.dtypes
```

```python
cleaned_data["country"].str.contains("india", case=False)
```

```python
cleaned_data["tax"] = cleaned_data["income"].apply(lambda x: "20%" if x >= 50000 else "10%")
cleaned_data
```

```python
gender_map = {"male":"M", "female": "F", "unknown": "U", "NaN": "U"}
cleaned_data["gender"] = cleaned_data["gender"].str.lower().map(gender_map)
cleaned_data
```

```python
cleaned_data = cleaned_data.assign(new_inc = cleaned_data["income"] * 1.1)
cleaned_data
```

```python
cleaned_data["income"] = cleaned_data["income"].fillna(round(cleaned_data["income"].mean()))
cleaned_data["gender"] = cleaned_data["gender"].fillna("U")
cleaned_data["name"] = cleaned_data["name"].fillna("Unknown")
cleaned_data
```

```python
cleaned_data["country"] = cleaned_data["country"].replace("USA", "US")
cleaned_data
```

```python
cleaned_data.columns = cleaned_data.columns.str.capitalize()
cleaned_data = cleaned_data.rename(columns={"Income": "Salary"})
cleaned_data
```

```python
cleaned_data = cleaned_data.rename(columns={"New_inc": "New Income"})
```

```python
cleaned_data.sort_values(["Salary", "Age"], ascending=False)
```

```python
cleaned_data.sort_index()
```

```python
cleaned_data = cleaned_data.sort_values(["Salary", "Age"], ascending=False)
# cleaned_data.reset_index()
cleaned_data.reset_index(drop=True)
```

```python
cleaned_data["Ranking"] = cleaned_data["Salary"].rank(ascending=False)
cleaned_data
```

```python
cleaned_data[['Id', 'Name', 'Age', 'Country', 'Gender', 'Salary', 'New Income',
       'Ranking', 'Tax']]
```

```python
new_col_order = [col for col in cleaned_data.columns if col != "Id"]
new_col_order.append("Id")
cleaned_data[new_col_order]
```

```python
cleaned_data.to_csv("Data-Files/cleaned_data.csv")
```

```python
cleaned_data.groupby("Country").agg({
    "New Income": "max",
    "Age": "mean" 
})
```

```python
melted_data = cleaned_data.melt(
    id_vars=["Id", "Name", "Salary"],
    value_vars=["Country", "Gender", "Age"],
    var_name="Metrics",
    value_name="value"
)


melted_data.pivot(
    index=["Id", "Name", "Salary"],
    columns="Metrics",
    values="value"
)
```
