```python
arr2D = np.arange(1, 10).reshape(3, 3)
print(arr2D)

print(np.sum(arr2D))
print("Sum of all the columns: ", np.sum(arr2D, axis=0))
print("Sum of all the rows:\t", np.sum(arr2D, axis=1))


print(arr2D[0:2, 1:2])
```

```python
arr = np.arange(1, 5).reshape(2, 2)
print(np.mean(arr), np.std(arr))

normalized_arr = (arr - np.mean(arr)) / np.std(arr)
print(normalized_arr, np.mean(normalized_arr), np.std(arr))
```
