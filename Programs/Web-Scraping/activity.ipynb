```python
import pandas as pd 
from bs4 import BeautifulSoup
import requests
import time
```

```python
i = 1
flag = True

while flag:
    URL = f"https://quotes.toscrape.com/page/{i}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text, "lxml")
    data = soup.find("div", class_="col-md-8")
    data = data.find_next("div", class_="col-md-8")

    clean_text = data.contents[0].strip()

    if clean_text =="No quotes found!":
        flag = False
    else:
        with open("../Data-Files/quotes_data.html", "a") as file:
            file.write(str(data))
    i += 1
    time.sleep(1)
    
```

```python
with open("../Data-Files/quotes_data.html", 'r') as file:
    data = file.read()

soup = BeautifulSoup(data, "lxml")
```

```python
author = []
text = []
all_tags = soup.find_all("div", class_="tags")
for tag in all_tags:
    all_a = tag.find_all("a", class_="tag")
    for a in all_a:
        if a.get_text() == "life":
            span_author = tag.find_previous_sibling()
            author.append(span_author.get_text().replace("(about)", "").replace("by", "").replace("\n", "").strip())

            span_text = span_author.find_previous_sibling()
            text.append(span_text.get_text().replace("“", "").replace("”", "").strip())
```

```python
df = pd.DataFrame({
    "Quote": text,
    "Author": author
})
```

```python
df.to_csv("../Data-Files/life_quotes.csv")
```
