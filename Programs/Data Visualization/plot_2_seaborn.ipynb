```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
```

```python
tips.head()
```

```python
tips.describe()
```

```python
tips.info()
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker", 
    size="size",
    style="smoker"
)
plt.show()
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker", 
    size="size",
    style="smoker",
    kind="line"
)
plt.show()
```

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

```python
flights = sns.load_dataset("flights")

sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)
```

```python
flights_pivot = flights.pivot(index="month", columns="year", values="passengers")
flights_pivot
```

```python
sns.heatmap(
    flights_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)
```

```python
fig, ax = plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    ax=ax,
    ci=None,
    errorbar=None
)
ax.set_title("Average Bill Value for Each Day")
ax.set_xlabel("Days")
ax.set_ylabel("Total Bill")
fig.tight_layout()
```
