```python
oscar_movies = [
    "The Dark Knight",
    "The hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012] 
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

plt.plot(years, oscar_revenue)
plt.title("Oscar movie in each year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.show()
```

```python
non_oscar_movies = ["Slimdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoin"]
non_oscar_revenue = [378, 2778, 829, 185, 275]


plt.plot(years, oscar_revenue,"o--r", label="Oscar Movies", linewidth=2)
plt.plot(years, non_oscar_revenue, "v--", label="Non Oscar Movie", color="#4917DF")
plt.title("Oscar movie VS Non Oscar movie [In Renvenue]")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()
plt.show()
```

```python
with plt.xkcd():

    plt.plot(years, oscar_revenue, label="Oscar Movies")
    plt.plot(years, non_oscar_revenue,  label="Non Oscar Movie")
    plt.title("Oscar movie VS Non Oscar movie [In Renvenue]")
    plt.xlabel("Years")
    plt.ylabel("Revenue (in $M)")
    plt.legend()
    # plt.grid()
    plt.show()
```

```python
plt.plot(oscar_movies, oscar_revenue)
plt.title("Movie Revenue")
plt.xlabel("Movies")
plt.ylabel("Revenue [In $M]")
plt.tight_layout()
plt.savefig("../Data-Files/Movie_Revenue.png")
plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
cities = ["New York", "London", "Delhi", "Tokyo"]
temperatures = [
    [22, 23, 21, 24, 25], # New York
    [18, 19, 17, 20, 21], # London
    [30, 32, 31, 33, 34], # Delhi
    [25, 26, 24, 27, 28]  # Tokyo
]

fig, ax = plt.subplots(2 , 2, figsize=(14, 7))

ax[0][0].plot(days, temperatures[0])
ax[0][0].set_xlabel("Days")
ax[0][0].set_ylabel("Temperature")
ax[0][0].set_title("Temperature of New York")

ax[0][1].plot(days, temperatures[1])
ax[0][1].set_xlabel("Days")
ax[0][1].set_ylabel("Temperature")
ax[0][1].set_title("Temperature of London")

ax[1][0].plot(days, temperatures[2])
ax[1][0].set_xlabel("Days")
ax[1][0].set_ylabel("Temperature")
ax[1][0].set_title("Temperature of Delhi")

ax[1][1].plot(days, temperatures[3])
ax[1][1].set_xlabel("Days")
ax[1][1].set_ylabel("Temperature")
ax[1][1].set_title("Temperature of Tokyo")

fig.suptitle("Temperatures of Cities")
fig.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(2 , 2, figsize=(13, 7))

def deploy_plots(i, j, sr_no):
    ax[i][j].plot(days, temperatures[sr_no], marker="o")
    ax[i][j].set_title(f"Temperature of {cities[sr_no]}")
    ax[i][j].grid(True)

for i in range(2):
    for j in range(2):
        sr_no = int(str(i) + str(j), 2)
        deploy_plots(i, j, sr_no)

fig.suptitle("Temperatures of Cities")
fig.supxlabel("Days")
fig.supylabel("Temperature")
fig.tight_layout()
plt.show()
        
```

```python
fig, axes = plt.subplots(2 , 2, figsize=(13, 7))

axes = axes.flatten()

for i, ax in enumerate(axes):
    ax.plot(days, temperatures[i], marker="o")
    ax.set_title(f"Temperature of {cities[i]}")
    ax.grid(True)


fig.suptitle("Temperatures of Cities")
fig.supxlabel("Days")
fig.supylabel("Temperature")
fig.tight_layout()
plt.show()
    
```
