```python
oscar_movies = [
    "The Dark Knight",
    "The hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012] 
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

plt.plot(years, oscar_revenue)
plt.title("Oscar movie in each year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.show()
```

```python
non_oscar_movies = ["Slimdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoin"]
non_oscar_revenue = [378, 2778, 829, 185, 275]


plt.plot(years, oscar_revenue,"o--r", label="Oscar Movies", linewidth=2)
plt.plot(years, non_oscar_revenue, "v--", label="Non Oscar Movie", color="#4917DF")
plt.title("Oscar movie VS Non Oscar movie [In Renvenue]")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()
plt.show()
```

```python
with plt.xkcd():

    plt.plot(years, oscar_revenue, label="Oscar Movies")
    plt.plot(years, non_oscar_revenue,  label="Non Oscar Movie")
    plt.title("Oscar movie VS Non Oscar movie [In Renvenue]")
    plt.xlabel("Years")
    plt.ylabel("Revenue (in $M)")
    plt.legend()
    # plt.grid()
    plt.show()
```

```python
plt.plot(oscar_movies, oscar_revenue)
plt.title("Movie Revenue")
plt.xlabel("Movies")
plt.ylabel("Revenue [In $M]")
plt.tight_layout()
plt.savefig("../Data-Files/Movie_Revenue.png")
plt.show()
```

```python
plt.bar(years, oscar_revenue)
# plt.bar(years, non_oscar_revenue)
plt.title("Revenue in Each Year")
plt.xlabel("Years")
plt.ylabel("Revenue [In $M]")
for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i] + 10, str(oscar_revenue[i]), ha="center")

plt.ylim(0, max(oscar_revenue) + 100)
# plt.xlim(min(years), max(years) + 100)
plt.show()
```

```python
x = np.arange(len(years))
width = 0.4

plt.title("Oscar VS Non-Oscar Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (In $M)")
plt.bar(x-width / 1.8, oscar_revenue, width, label="Oscar Movies")
plt.bar(x+width / 1.8, oscar_revenue, width, label="Oscar [1] Movies")
# plt.bar(x + width / 0.8, non_oscar_revenue, width, label="Non-Oscar Movies")

plt.xticks(x, years)
plt.legend()
plt.show()
```

```python
plt.title("Movie Revenue")
plt.barh(oscar_movies, oscar_revenue)
plt.ylabel("Movies")
plt.xlabel("Revenue")
```

```python
people = [f"Person {chr(Alphabet)}" for Alphabet in range(ord("A"), ord("K"))]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [i for i in range(110, 155, 5)]
bp.append(122)
bp.sort()
bp[bp.index(140)] = 123
bp
```

```python
colors = ["green" if x < 135 else "red" for x in bp]
# sizes = 
plt.scatter(age, bp, cmap="Purples", c=bp ,s=bp,)
plt.title("Age VS Blood-pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar(label="BP")
for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]), xytext=(age[i] + 1, bp[i]))
plt.xlim(min(age), max(age) + 10)
plt.ylim(min(bp) -10, max(bp)+5)
plt.show()
```

```python
city = [f"City {chr(alphabet)}" for alphabet in range(ord("A"), ord("F"))]
winter_temp = [random.randint(0, 11) for _ in range(5)]
winter_humidity = [random.randint(70, 86) for _ in range(5)]

summer_temp = [random.randint(25, 36) for _ in range(5)]
summer_humidity = [random.randint(45, 65) for _ in range(5)]
```

```python
plt.scatter(winter_temp, winter_humidity, label="Winter")
plt.scatter(summer_temp, summer_humidity, label="Summer")

plt.title("Summer VS Winter")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity %")
plt.legend()

plt.show()
```
