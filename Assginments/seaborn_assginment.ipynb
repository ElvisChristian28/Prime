```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)
plt.grid(True)
```

### Interpret the Data
- Identify which day has the highest median bill: Sun by men
- Outliers are mostly of sat and sun, maybe because of holiday

```python
fig, ax = plt.subplots(1 , 1)

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker",
    ax=ax
)

ax.set_title("total_bill vs tip")
```
